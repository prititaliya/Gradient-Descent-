```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import make_regression
x=make_regression(n_features=1,n_targets=1,n_samples=4,noise=80)
```

```python
y=pd.DataFrame(x[1])
x=pd.DataFrame(x[0])
```

```python
plt.scatter(x,y)
plt.show()
```

```python
from sklearn.linear_model import LinearRegression
lr=LinearRegression()
lr.fit(x,y)
```

```python
correct_m=lr.coef_
correct_b=lr.intercept_
correct_b
```

```python
plt.scatter(x,y)
plt.plot(x,correct_m*x+correct_b,color='red')
plt.show()
```

```python
rand_b=10
plt.scatter(x,y)
plt.plot(x,correct_m*x+rand_b,color='red')
plt.plot(x,correct_m*x+correct_b,color='green')
plt.show()
```

- lets start gradient descent with randome b starts with 10 and m is from lr (we assuming that we already know correct m ).

```python
loss_function_slop=-2*(np.sum(y-correct_m*x-rand_b))
loss_function_slop
```

```python
lr=0.01
step_size=loss_function_slop*lr
step_size
```

```python
new_b=correct_b-step_size
new_b
```

```python
y_pred1=((correct_m*x)+new_b)
plt.scatter(x,y)
plt.plot(x,y_pred1,color='red')
plt.plot(x,correct_m*x+rand_b,color='green')
plt.plot(x,correct_m*x+correct_b,color='blue')
plt.show()
```

```python
loss_function_slop=-2*(np.sum(y-correct_m*x-new_b))
step_size=loss_function_slop*lr
new_b=new_b-step_size
y_pred2=((correct_m*x)+new_b)
plt.scatter(x,y)
plt.plot(x,y_pred2,color='red')
plt.plot(x,correct_m*x+rand_b,color='green')
plt.plot(x,correct_m*x+correct_b,color='blue')
plt.show()
```

```python
b=100
m=correct_m
lr = 0.01
epochs=100

for i in range(epochs):
    loss_function_slop=-2*(np.sum(y-m*x-b))
    step_size=loss_function_slop*lr
    b=b-step_size
    y_pred=m*x+b
    plt.scatter(x,y)
    plt.plot(x,y_pred,color='red')
    plt.plot(x,correct_m*x+rand_b,color='green')
```

```python
b,correct_b
```

```python
class gradient_descent():
    def __init__(self, learning_rate, epochs):
        self.learning_rate = learning_rate
        self.epochs = epochs
        self.m=correct_m
        self.b=10

    def fit(self, x, y):
        for i in range(self.epochs):
            loss_function=(-2)*(y-(self.m*x+self.b))
            self.b=self.b-(self.learning_rate*loss_function);

        return self.m,self.b
```

```python
gd=gradient_descent(0.01,100)
new_m,new_b=gd.fit(x,y)
```
